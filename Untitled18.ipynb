```python
"""
Insurance Operations Data Generator
Simulates 2 years of daily voice & non-voice channel data
with realistic capacity constraints and backlog accumulation.
"""

import pandas as pd
import numpy as np

# ============================================================
# CONFIGURATION (change these to experiment)
# ============================================================
START_DATE = '2023-01-01'
END_DATE   = '2024-12-31'
RANDOM_SEED = 42

# Team size (FTE = Full-Time Equivalent employees)
TOTAL_FTE_AVAILABLE = 70          # 70 people in operations team
PRODUCTIVE_HOURS_PER_FTE = 7.5    # 7.5 productive hours/day (breaks, meetings excluded)

# Workload assumptions
VOICE_BASE_CALLS = 1200           # average calls/day
VOICE_AHT_MEAN = 5.5              # minutes per call
NON_VOICE_BASE_TICKETS = 800      # average tickets/day
MINUTES_PER_TICKET = 30           # 0.5 hours per ticket

# ============================================================
# STEP 1: GENERATE DATE RANGE
# ============================================================
dates = pd.date_range(start=START_DATE, end=END_DATE, freq='D')
np.random.seed(RANDOM_SEED)

# ============================================================
# STEP 2: SIMULATE RAW VOLUMES WITH REALISTIC PATTERNS
# ============================================================
# Base volumes with random noise
voice_calls = np.random.poisson(lam=VOICE_BASE_CALLS, size=len(dates))
voice_aht = np.random.normal(loc=VOICE_AHT_MEAN, scale=0.8, size=len(dates))
non_voice_tickets = np.random.poisson(lam=NON_VOICE_BASE_TICKETS, size=len(dates))
non_voice_resolution = np.random.normal(loc=2.5, scale=0.5, size=len(dates))

# --- SEASONALITY ---
month = pd.DatetimeIndex(dates).month
seasonality = np.ones(len(dates))
seasonality[np.isin(month, [11, 12])] = 1.30   # Q4 spike (+30%)
seasonality[np.isin(month, [7, 8])]   = 0.80   # Summer dip (−20%)

# --- DAY-OF-WEEK EFFECT (Mondays are busy) ---
dayofweek = pd.DatetimeIndex(dates).dayofweek   # 0 = Monday
dow_factor = np.where(dayofweek == 0, 1.20,          # Monday +20%
              np.where(dayofweek == 5, 0.60,          # Saturday −40%
              np.where(dayofweek == 6, 0.50, 1.00)))  # Sunday −50%

# --- MONTH-END EFFECT (billing, reporting peaks) ---
day_of_month = pd.DatetimeIndex(dates).day
month_end_factor = np.where(day_of_month > 25, 1.15, 1.00)

# Apply all multipliers
voice_calls = (voice_calls * seasonality * dow_factor * month_end_factor).astype(int)
non_voice_tickets = (non_voice_tickets * seasonality * dow_factor * month_end_factor).astype(int)

# ============================================================
# STEP 3: CALCULATE WORKLOAD (HOURS)
# ============================================================
voice_workload_hours = (voice_calls * voice_aht) / 60
non_voice_workload_hours = (non_voice_tickets * MINUTES_PER_TICKET) / 60
total_workload_hours = voice_workload_hours + non_voice_workload_hours

# ============================================================
# STEP 4: CALCULATE CAPACITY (WITH REALISTIC VARIATION)
# ============================================================
# Not every FTE is available every day (sick, training, holidays)
# Capacity varies between 65% and 100% of nominal team size
capacity_variation = np.random.normal(loc=0.90, scale=0.08, size=len(dates))
capacity_variation = np.clip(capacity_variation, 0.65, 1.00)

capacity_fte = (TOTAL_FTE_AVAILABLE * capacity_variation).round(1)
daily_capacity_hours = capacity_fte * PRODUCTIVE_HOURS_PER_FTE

# FTE required to handle the workload
fte_required = total_workload_hours / PRODUCTIVE_HOURS_PER_FTE

# FTE gap: positive = understaffed, negative = overstaffed
fte_gap = fte_required - capacity_fte

# ============================================================
# STEP 5: SIMULATE BACKLOG & LEAD TIME
# ============================================================
backlog = 0.0
backlog_list = []
lead_time_list = []

for i in range(len(dates)):
    workload = total_workload_hours.iloc[i] if hasattr(total_workload_hours, 'iloc') else total_workload_hours[i]
    cap_hours = daily_capacity_hours[i]
    
    if workload > cap_hours:
        # Can't finish today → backlog grows
        backlog += workload - cap_hours
    else:
        # Finish today's work + reduce backlog
        backlog = max(0, backlog - (cap_hours - workload))
    
    backlog_list.append(backlog)
    
    # Lead time = how many days to clear current backlog
    lead_time = backlog / cap_hours if cap_hours > 0 else 0
    lead_time_list.append(lead_time)

# ============================================================
# STEP 6: BUILD FINAL DATAFRAME
# ============================================================
df = pd.DataFrame({
    'date': dates,
    'voice_calls': voice_calls,
    'voice_aht_min': voice_aht.round(2),
    'non_voice_tickets': non_voice_tickets,
    'non_voice_resolution_days': non_voice_resolution.round(2),
    'voice_workload_hours': voice_workload_hours.round(2),
    'non_voice_workload_hours': non_voice_workload_hours.round(2),
    'total_workload_hours': total_workload_hours.round(2),
    'capacity_fte': capacity_fte,
    'daily_capacity_hours': daily_capacity_hours.round(2),
    'fte_required': fte_required.round(2),
    'fte_gap': fte_gap.round(2),
    'backlog': [round(b, 2) for b in backlog_list],
    'lead_time_days': [round(l, 2) for l in lead_time_list]
})

# ============================================================
# STEP 7: SAVE & VALIDATE
# ============================================================
df.to_csv('insurance_operations_data.csv', index=False)

print("=" * 60)
print("DATA GENERATION SUMMARY")
print("=" * 60)
print(f"Date range: {df['date'].min().date()} → {df['date'].max().date()}")
print(f"Total days: {len(df)}")
print(f"Team size (nominal): {TOTAL_FTE_AVAILABLE} FTE")
print(f"Productive hours/FTE/day: {PRODUCTIVE_HOURS_PER_FTE}")
print()
print(f"Avg daily workload: {df['total_workload_hours'].mean():.0f} hours")
print(f"Avg daily capacity: {df['daily_capacity_hours'].mean():.0f} hours")
print(f"Avg FTE required: {df['fte_required'].mean():.1f}")
print(f"Avg FTE available: {df['capacity_fte'].mean():.1f}")
print()
print(f"Days with backlog > 0: {(df['backlog'] > 0).sum()} / {len(df)}")
print(f"Max backlog: {df['backlog'].max():.0f} hours")
print(f"Avg lead time: {df['lead_time_days'].mean():.2f} days")
print(f"Max lead time: {df['lead_time_days'].max():.2f} days")
print()
print(f"Days understaffed (gap > 0): {(df['fte_gap'] > 0).sum()} / {len(df)}")
print(f"Days overstaffed (gap < 0): {(df['fte_gap'] < 0).sum()} / {len(df)}")
print()
print("First 5 rows:")
print(df.head().to_string())
print()
print("Sample of busy days (backlog > 0):")
print(df[df['backlog'] > 0].head(10).to_string())
```

```python
"""
Run all KPI queries and export results as CSV files
for Power BI consumption.
"""

import pandas as pd
import sqlite3

conn = sqlite3.connect('insurance_ops.db')

# Define queries inline (or read from kpi_queries.sql)
queries = {
    'kpi_monthly': """
        SELECT strftime('%Y-%m', date) AS month,
               ROUND(AVG(voice_calls), 0) AS avg_daily_calls,
               ROUND(AVG(total_workload_hours), 0) AS avg_workload_hrs,
               ROUND(AVG(fte_required), 1) AS avg_fte_required,
               ROUND(AVG(capacity_fte), 1) AS avg_fte_available,
               ROUND(AVG(fte_gap), 1) AS avg_fte_gap,
               ROUND(MAX(backlog), 0) AS max_backlog,
               ROUND(AVG(lead_time_days), 2) AS avg_lead_time
        FROM daily_operations
        GROUP BY strftime('%Y-%m', date)
        ORDER BY month
    """,
    'kpi_weekly': """
        SELECT strftime('%Y-W%W', date) AS week,
               ROUND(AVG(total_workload_hours), 0) AS avg_workload,
               ROUND(AVG(daily_capacity_hours), 0) AS avg_capacity,
               ROUND(MAX(backlog), 0) AS peak_backlog,
               ROUND(AVG(lead_time_days), 2) AS avg_lead_time
        FROM daily_operations
        GROUP BY strftime('%Y-W%W', date)
        ORDER BY week
    """,
    'kpi_channel_split': """
        SELECT strftime('%Y-%m', date) AS month,
               ROUND(SUM(voice_workload_hours), 0) AS voice_hours,
               ROUND(SUM(non_voice_workload_hours), 0) AS non_voice_hours,
               ROUND(100.0 * SUM(voice_workload_hours) /
                     SUM(total_workload_hours), 1) AS voice_pct
        FROM daily_operations
        GROUP BY strftime('%Y-%m', date)
        ORDER BY month
    """,
    'kpi_sla_breach': """
        SELECT strftime('%Y-%m', date) AS month,
               COUNT(*) AS total_days,
               SUM(CASE WHEN lead_time_days > 2 THEN 1 ELSE 0 END) AS breach_days,
               ROUND(100.0 * SUM(CASE WHEN lead_time_days > 2 THEN 1 ELSE 0 END)
                     / COUNT(*), 1) AS breach_pct
        FROM daily_operations
        GROUP BY strftime('%Y-%m', date)
        ORDER BY month
    """,
    'kpi_worst_days': """
        SELECT date, ROUND(backlog, 0) AS backlog,
               ROUND(lead_time_days, 2) AS lead_time,
               ROUND(fte_gap, 1) AS fte_gap
        FROM daily_operations
        ORDER BY backlog DESC
        LIMIT 10
    """,
    'kpi_quarterly': """
        SELECT CASE
                 WHEN strftime('%m', date) IN ('01','02','03') THEN 'Q1'
                 WHEN strftime('%m', date) IN ('04','05','06') THEN 'Q2'
                 WHEN strftime('%m', date) IN ('07','08','09') THEN 'Q3'
                 ELSE 'Q4'
               END || ' ' || strftime('%Y', date) AS quarter,
               ROUND(AVG(total_workload_hours), 0) AS avg_workload,
               ROUND(AVG(fte_gap), 1) AS avg_fte_gap,
               ROUND(MAX(backlog), 0) AS peak_backlog,
               ROUND(AVG(lead_time_days), 2) AS avg_lead_time
        FROM daily_operations
        GROUP BY strftime('%Y', date),
                 CASE
                   WHEN strftime('%m', date) IN ('01','02','03') THEN 'Q1'
                   WHEN strftime('%m', date) IN ('04','05','06') THEN 'Q2'
                   WHEN strftime('%m', date) IN ('07','08','09') THEN 'Q3'
                   ELSE 'Q4'
                 END
        ORDER BY quarter
    """
}

# Run and export
for name, query in queries.items():
    result = pd.read_sql(query, conn)
    result.to_csv(f'{name}.csv', index=False)
    print(f"\n{'=' * 60}")
    print(f"📊 {name.upper()}")
    print('=' * 60)
    print(result.to_string(index=False))

conn.close()
print("\n✅ All KPI tables exported as CSV for Power BI")
```

```python
"""
Insurance Operations - Workload Forecasting
Predicts next 30 days of workload → converts to FTE requirement
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.holtwinters import ExponentialSmoothing
import warnings
warnings.filterwarnings('ignore')

plt.rcParams['figure.figsize'] = (15, 5)

# ============================================================
# 1. LOAD DATA
# ============================================================
df = pd.read_csv('insurance_operations_data.csv')
df['date'] = pd.to_datetime(df['date'], dayfirst=True, format='mixed', errors='coerce')
df = df.sort_values('date').reset_index(drop=True)

print("=" * 60)
print("FORECASTING: WORKLOAD & FTE REQUIREMENTS")
print("=" * 60)
print(f"History: {df['date'].min().date()} → {df['date'].max().date()}")
print(f"Days of history: {len(df)}")

# ============================================================
# 2. TRAIN/TEST SPLIT (validate model on last 30 days)
# ============================================================
HOLDOUT_DAYS = 30
train = df.iloc[:-HOLDOUT_DAYS].copy()
test  = df.iloc[-HOLDOUT_DAYS:].copy()

print(f"\nTrain size: {len(train)} days")
print(f"Test size:  {len(test)} days (last 30 days)")

# ============================================================
# 3. BUILD HOLT-WINTERS MODEL
# ============================================================
# Use weekly seasonality (7 days) — captures Monday/Friday pattern
# Additive trend + additive seasonal (workload is in hours, no exponential growth)
ts_train = train.set_index('date')['total_workload_hours']

model = ExponentialSmoothing(
    ts_train,
    trend='add',
    seasonal='add',
    seasonal_periods=7,
    initialization_method='estimated'
)
fit = model.fit(optimized=True)

print(f"\n✅ Model trained")
print(f"   AIC: {fit.aic:.2f}")
print(f"   Smoothing level:     {fit.params['smoothing_level']:.4f}")
print(f"   Smoothing trend:     {fit.params['smoothing_trend']:.4f}")
print(f"   Smoothing seasonal:  {fit.params['smoothing_seasonal']:.4f}")

# ============================================================
# 4. VALIDATE ON TEST SET
# ============================================================
test_pred = fit.forecast(len(test))
test_actual = test['total_workload_hours'].values

# Metrics
mae = np.mean(np.abs(test_actual - test_pred))
rmse = np.sqrt(np.mean((test_actual - test_pred) ** 2))
mape = np.mean(np.abs((test_actual - test_pred) / test_actual)) * 100

print(f"\n📊 VALIDATION ON LAST 30 DAYS")
print(f"   MAE:  {mae:.1f} hours/day")
print(f"   RMSE: {rmse:.1f} hours/day")
print(f"   MAPE: {mape:.1f}%  (accuracy: {100-mape:.1f}%)")

# ============================================================
# 5. RETRAIN ON FULL DATA & FORECAST NEXT 30 DAYS
# ============================================================
ts_full = df.set_index('date')['total_workload_hours']

final_model = ExponentialSmoothing(
    ts_full,
    trend='add',
    seasonal='add',
    seasonal_periods=7,
    initialization_method='estimated'
)
final_fit = final_model.fit(optimized=True)

FORECAST_DAYS = 30
forecast = final_fit.forecast(FORECAST_DAYS)

# Build forecast dataframe
forecast_df = pd.DataFrame({
    'date': forecast.index,
    'forecasted_workload_hours': forecast.values.round(2)
})

# ============================================================
# 6. CONVERT FORECAST → FTE REQUIREMENT
# ============================================================
PRODUCTIVE_HOURS_PER_FTE = 7.5
CURRENT_FTE = 70

forecast_df['forecasted_fte_required'] = (
    forecast_df['forecasted_workload_hours'] / PRODUCTIVE_HOURS_PER_FTE
).round(1)

forecast_df['available_fte'] = CURRENT_FTE
forecast_df['fte_gap'] = (
    forecast_df['forecasted_fte_required'] - forecast_df['available_fte']
).round(1)

# Day of week for context
forecast_df['day_of_week'] = forecast_df['date'].dt.day_name()

# ============================================================
# 7. OUTPUT FORECAST SUMMARY
# ============================================================
print("\n" + "=" * 60)
print(f"📈 30-DAY FORECAST (starting {forecast_df['date'].min().date()})")
print("=" * 60)
print(forecast_df.to_string(index=False))

# ============================================================
# 8. HIRING RECOMMENDATION
# ============================================================
avg_fte_needed = forecast_df['forecasted_fte_required'].mean()
max_fte_needed = forecast_df['forecasted_fte_required'].max()
avg_gap = forecast_df['fte_gap'].mean()
max_gap = forecast_df['fte_gap'].max()

print("\n" + "=" * 60)
print("💡 HIRING RECOMMENDATION")
print("=" * 60)
print(f"Avg FTE required (next 30 days): {avg_fte_needed:.1f}")
print(f"Peak FTE required:               {max_fte_needed:.1f}")
print(f"Current FTE available:           {CURRENT_FTE}")
print(f"Avg FTE gap:                     {avg_gap:+.1f}")
print(f"Peak FTE gap:                    {max_gap:+.1f}")
print()

if avg_gap > 5:
    print(f"🚨 ACTION: Hire ~{int(np.ceil(avg_gap))} FTEs immediately")
elif avg_gap > 0:
    print(f"⚠️  ACTION: Add {int(np.ceil(avg_gap))} FTEs or overtime to cover gap")
else:
    print(f"✅ ACTION: No hiring needed — capacity sufficient")

# ============================================================
# 9. VISUALIZATION
# ============================================================
fig, ax = plt.subplots(figsize=(15, 6))

# Historical (last 90 days for context)
ax.plot(df['date'].iloc[-90:], df['total_workload_hours'].iloc[-90:],
        label='Historical workload', color='steelblue', lw=1.5)

# Forecast
ax.plot(forecast_df['date'], forecast_df['forecasted_workload_hours'],
        label='Forecast (30 days)', color='crimson', lw=2, linestyle='--')

# Capacity line
capacity_line = CURRENT_FTE * PRODUCTIVE_HOURS_PER_FTE
ax.axhline(y=capacity_line, color='darkorange', linestyle=':', lw=2,
           label=f'Current capacity ({CURRENT_FTE} FTE × 7.5h = {capacity_line:.0f}h)')

ax.set_title('Workload Forecast — Next 30 Days', fontsize=14, fontweight='bold')
ax.set_xlabel('Date'); ax.set_ylabel('Workload (hours)')
ax.legend(loc='upper left')
plt.tight_layout()
plt.savefig('08_forecast.png', dpi=100)
plt.show()

# FTE Gap forecast
fig, ax = plt.subplots(figsize=(15, 5))
colors = np.where(forecast_df['fte_gap'] > 0, 'red', 'green')
ax.bar(forecast_df['date'], forecast_df['fte_gap'], color=colors, width=0.8)
ax.axhline(y=0, color='black', lw=0.8)
ax.set_title('Forecasted FTE Gap — Next 30 Days (red = understaffed)',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Date'); ax.set_ylabel('FTE gap')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('09_forecast_fte_gap.png', dpi=100)
plt.show()

# ============================================================
# 10. EXPORT
# ============================================================
forecast_df.to_csv('forecast_next_30_days.csv', index=False)
print("\n✅ Forecast exported: forecast_next_30_days.csv")
print("✅ Charts saved: 08_forecast.png, 09_forecast_fte_gap.png")
```
